```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('imdb_top_1000.csv')
```

```python
df.head()
```

```python
df1 = df.drop(['Runtime','Released_Year','Certificate','Meta_score','No_of_Votes','IMDB_Rating','Gross'],axis=1)
```

```python
df1["tags"] = ( df1['Series_Title'] + ' ' + df1['Genre'] + ' ' + df1['Director'] + ' ' + df1['Star1'] + ' ' + df1['Star2'] + ' ' + df1['Star3'] + ' ' + df1['Star4'] )
```

```python
df1.head()
```

```python
df1['tags'][0]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizedText = TfidfVectorizer(stop_words='english',max_features=5000)
```

```python
finalVectors = vectorizedText.fit_transform(df1['tags'])
```

```python
finalVectors
```

```python
df1.shape
```

```python
df1.info()
```

```python
df1.head()
```

```python
from sklearn.neighbors import NearestNeighbors
```

```python
Knn = NearestNeighbors(n_neighbors=6,metric='cosine')
Knn.fit(finalVectors)
```

```python
def recomend(movie):
    indexes = df1[df1['Series_Title'] == movie].index[0]

    distance, indices = Knn.kneighbors(finalVectors[indexes])

    
    for i in indices[0][1:]:
        print(df1.iloc[i]['Series_Title'])

recomend('The Dark Knight')
```
